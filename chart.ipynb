```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
```

```python
x_vals = np.arange(20)
y_vals = [(1 + 2 * (1 - (min([10, x])/10))**2) for x in x_vals]
y_vals = [np.round(y, 2) for y in y_vals]

sat_y_vals = np.arange(0, 3.5, 0.1)
sat_x_vals = [10 for x in sat_y_vals]
```

```python
fig = plt.figure(dpi=100)
ax = fig.add_subplot(1, 1, 1)
ax.plot(x_vals, y_vals, label='Коефіцієнт посилення')
ax.plot(sat_x_vals, sat_y_vals, label='Момент насичення', linestyle='--')
ax.set_ylim(0, 3.5)
ax.set_title('Коефіцієнт посилення початкових взаємодій')
ax.set_xlabel('Номер взаємодії агента')
ax.set_ylabel('Значення коефіцієнту')
ax.legend(loc='best')

ax.set_xticks([])
ax.set_yticks([])
ax.set_xticklabels([])
ax.set_yticklabels([])

fig.show()
```
